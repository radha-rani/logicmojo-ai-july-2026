## Reading input

`input()` reads a line the user types and returns it. It blocks the program while waiting for the keyboard, so we cannot run a live `input()` here without hanging the notebook. Instead we use the string it *would* return, to show the one rule that trips everyone up.

```python
# Take input from user
qty = input("How many? ")

print(type(qty))          # <class 'str'>, input() ALWAYS returns a string, even for digits

# doing math on the raw value is a trap:
print(qty * 2)            # string repetition

qty = int(qty)            # convert first
price = 4.5
print("total:", qty * price)
```

## Tuples

Ordered and immutable: a fixed record you can unpack. (Returning several values from a function is a common use; we get to that once functions arrive.)

```python
point = (4, 9)             # an ordered, fixed pair
x, y = point               # unpack into variables
print(x, y)                # 4 9
print(point[0])            # 4

rolls = (6, 1, 6, 3, 6)
print(rolls.count(6))      # 3  -> how many sixes
print(rolls.index(3))      # 3  -> position of the first 3
```

## Sets

Unique items, fast membership, and set algebra. (Sets are unordered, so we `sorted()` the results for stable output.)

```python
visits = ["ada", "bob", "ada", "cy", "bob"]
print(len(set(visits)), "unique visitors")   # dedupe
```

```python
monday = {"ada", "bob", "cy"}
tuesday = {"bob", "cy", "dee"}

print(sorted(monday & tuesday))   # both days
print(sorted(monday | tuesday))   # either day
print(sorted(monday - tuesday))   # only monday
```

```python
monday & tuesday
```

```python
# empty set

s = set()  # empty set, not {}, which is an empty dict
```

## List comprehension

Map, filter with a trailing `if`, and choose values with a conditional expression.

```python
prices = [40, 120, 250, 80]

print([p * 0.9 for p in prices])                       # map: 10% off
print([p for p in prices if p > 100])                  # filter
print(["premium" if p > 100 else "standard" for p in prices])  # conditional
```

## Dict comprehension

Same shape, building `key: value`, and it can filter too.

```python
squares = {n: n * n for n in range(1, 6)}
print(squares)

big = {n: n * n for n in range(1, 6) if n * n > 5}
print(big)
```

## Functions

Inputs in, a result out with `return`. A default value fills in when you omit an argument. One rule: parameters with defaults must come **after** those without, so `def tax(amount, rate=0.1)` is valid but `def tax(rate=0.1, amount)` is a `SyntaxError`.

```python
def celsius_to_f(c):
    return c * 9 / 5 + 32

print(celsius_to_f(0))
print(celsius_to_f(100))
```

```python
def tax(amount, rate=0.1):     # rate has a default, and comes last
    return amount * rate

print(tax(200))        # uses the default rate
print(tax(200, 0.2))   # override it
```

**print versus return.** A function that only prints hands back `None`, so the caller has nothing to use.

```python
def show_total(items):
    print(sum(items))     # prints, but returns None

result = show_total([1, 2, 3])
print("result is:", result)
```

**Return more than one value.** Pack them into a tuple and unpack at the call site (this is where the tuple from earlier pays off).

```python
def min_max(nums):
    return min(nums), max(nums)   # a tuple: (low, high)

low, high = min_max([4, 9, 2, 7])   # unpack the returned tuple
print(low, high)
```

## Any number of arguments: `*args` and `**kwargs`

`*args` collects extra positional arguments into a tuple; `**kwargs` collects extra keyword arguments into a dict.

```python
def mean(*nums):              # extra positionals -> tuple
    return sum(nums) / len(nums)

print(mean(10, 20))           # 15.0
print(mean(4, 8, 12, 16))     # 10.0

def make_user(**fields):      # extra keywords -> dict
    if "name" in fields:
        print("Hello", fields["name"].title())
    return fields

u = make_user(name="Ada", role="admin", active=True)
print(u)
print(u["role"])
```

```python
make_user(role="Carpenter", Salary=500)
```

```python
def f1(a, b):
    def f2(x, y):
        return x * y
    return f2(a, b)   # return the result of f2
```

```python
f1(9, 10)
```

## Light type hints

A note about what goes in and comes out. It documents intent; it does not change behavior.

```python
def repeat(text: str, times: int = 2) -> str:
    return text * times

print(repeat("ab"))
print(repeat("hi", 3))
```

```python
repeat(4, 5)
```

## Build: three functions over a list of dicts

A common shape in real code: a list of records (dicts). Write small, type-hinted functions over it. The `assert` lines are quick checks; if they all pass, your functions behave.

```python
students = [
    {"name": "Ada", "score": 91},
    {"name": "Alan", "score": 72},
    {"name": "Grace", "score": 85},
    {"name": "Kay", "score": 60},
]

def total_score(rows: list) -> int:
    return sum(row["score"] for row in rows)

def passing(rows: list, cutoff: int = 70) -> list:
    return [row for row in rows if row["score"] >= cutoff]

def summary(rows: list) -> dict:
    total = total_score(rows)
    return {"count": len(rows), "total": total, "average": round(total / len(rows), 1)}

print("total:", total_score(students))
print("passing:", [row["name"] for row in passing(students)])
print("summary:", summary(students))

assert total_score(students) == 308
assert len(passing(students)) == 3
assert summary(students) == {"count": 4, "total": 308, "average": 77.0}
print("all checks passed")
```
