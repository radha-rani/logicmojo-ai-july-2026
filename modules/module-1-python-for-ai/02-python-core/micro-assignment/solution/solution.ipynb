## Problem 1: Two small functions

```python
def rectangle_area(width: int, height: int) -> int:
    return width * height

def greet(name: str, greeting: str = "Hello") -> str:
    return f"{greeting}, {name}"

print(f"area of 3 x 4 = {rectangle_area(3, 4)}")
print(greet("Ada"))
print(greet("Alan", "Hi"))
```

## Problem 2: Clean a messy record

```python
raw = "  Ada Lovelace,  ADA@Math.ORG , London  "

name, email, city = [field.strip() for field in raw.split(",")]
email = email.lower()

print(f"name: {name}")
print(f"email: {email}")
print(f"city: {city}")

domain = email.split("@")[1]
print(f"domain: {domain}")

initials = ".".join(part[0].upper() for part in name.split()) + "."
print(f"initials: {initials}")
```

## Problem 3: Scores dictionary

```python
scores = {"Ada": 91, "Alan": 72, "Grace": 85}

for name, score in scores.items():
    print(f"{name}: {score}")

average = sum(scores.values()) / len(scores)
print(f"average: {average:.2f}")
```

## Problem 4: Keep the long words

```python
words = ["ai", "python", "ml", "agent", "rag"]

kept = [w.upper() for w in words if len(w) > 2]
print(kept)
print(f"{len(kept)} words kept")
```

## Problem 5: Shared tags

```python
monday = ["python", "git", "venv", "ml"]
tuesday = ["git", "ml", "numpy", "pandas"]

both = sorted(set(monday) & set(tuesday))
either = sorted(set(monday) | set(tuesday))

print(f"in both: {both}")
print(f"in either: {either}")
print(f"{len(both)} shared, {len(either)} total")
```

## Problem 6: Longest name, without max()

```python
names = ["Ada", "Alan", "Grace", "Katherine", "Ed"]

lengths = {name: len(name) for name in names}
print(lengths)

longest = names[0]
for name in names:
    if len(name) > len(longest):
        longest = name
print(f"longest: {longest} ({len(longest)} letters)")
```
