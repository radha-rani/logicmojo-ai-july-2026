## Problem 1: Two small functions

```
area of 3 x 4 = 12
Hello, Ada
Hi, Alan
```

````python
# TODO: write two type-hinted functions, then call them for the expected output.
# 1. rectangle_area(width, height) -> returns width * height
# 2. greet(name, greeting="Hello") -> returns "<greeting>, <name>" (default greeting is Hello)
# Then print: the area of 3 x 4, greet("Ada"), and greet("Alan", "Hi").

# your code here

def rectangle_area(width: float, height: float) -> float:
    return width * height

def greet(name: str, greeting: str = "Hello") -> str:
    return f"{greeting}, {name}"


print(f"area of 3 x 4 = {rectangle_area(3, 4)}")
print(greet("Ada"))
print(greet("Alan", "Hi"))
````

## Problem 2: Clean a messy record

```
name: Ada Lovelace
email: ada@math.org
city: London
domain: math.org
initials: A.L.
```

````python
raw = "  Ada Lovelace,  ADA@Math.ORG , London  "

# TODO:
# 1. Split raw on commas and strip each field into name, email, city.
# 2. Print "name: <name>", "email: <lowercased email>", "city: <city>".
# 3. Print "domain: <part after @>" (split the email on "@").
# 4. Print "initials: <A.L.>": first letter of each name part, uppercased,
#    joined with "." and ending in a "." (e.g. "Ada Lovelace" -> "A.L.").

# your code here
name, email, city = [part.strip() for part in raw.split(",")]

print(f"name: {name}")
email = email.lower()
print(f"email: email")
print(f"city: {city}")

print(f"domain: {email.split('@')[1]}")
initials = ".".join(part[0] for part in name.split())+"."
print(f"Initials: {initials}")
````

## Problem 3: Scores dictionary

```
Ada: 91
Alan: 72
Grace: 85
average: 82.67
```

````python
scores = {"Ada": 91, "Alan": 72, "Grace": 85}

# TODO: print "name: score" for each entry (iterate with .items()),
# then print the average score to two decimals, like "average: 82.67".

# your code here
total = 0
cnt = 0
for key, value in scores.items():
    print(f"{key}: {value}")
    total += value
    cnt += 1

average = total/cnt
print(f"Average:{average:.2f}")
````

## Problem 4: Keep the long words

```
['PYTHON', 'AGENT', 'RAG']
3 words kept
```

````python
words = ["ai", "python", "ml", "agent", "rag"]

# TODO: with a list comprehension, build a list of the UPPERCASED words
# that are longer than two letters. Print the list, then "<n> words kept".

# your code here
uppercase_words = [word.upper() for word in words if len(word)>2]
print(uppercase_words)

print(f"{len(uppercase_words)} words kept")

````

## Problem 5: Shared tags

```
in both: ['git', 'ml']
in either: ['git', 'ml', 'numpy', 'pandas', 'python', 'venv']
2 shared, 6 total
```

````python
monday = ["python", "git", "venv", "ml"]
tuesday = ["git", "ml", "numpy", "pandas"]

# TODO: using sets, find the tags in BOTH lists and the tags in EITHER list.
# Print each as a sorted list, then "<shared> shared, <total> total".
# Hint: set(a) & set(b) is the intersection; set(a) | set(b) is the union.

# your code here
both = set(monday)&set(tuesday)
either = set(monday)|set(tuesday)
print(sorted(both))
print(sorted(either))
print(f"{len(both)} shared, {len(either)} total")
````

## Problem 6: Longest name, without max()

```
{'Ada': 3, 'Alan': 4, 'Grace': 5, 'Katherine': 9, 'Ed': 2}
longest: Katherine (9 letters)
```

````python
names = ["Ada", "Alan", "Grace", "Katherine", "Ed"]

# TODO:
# 1. Build a dict comprehension mapping each name to its length; print it.
# 2. Without using max(), find the longest name (scan and keep the best so far).
#    Print: "longest: <name> (<n> letters)".

# your code here

names_dict = {name:len(name) for name in names}
print(names_dict)

longest = names[0]
for name in names:
    if len(name)>len(longest):
        longest = name

print(f"Longest: {longest} ({len(longest)} letters)")
````
