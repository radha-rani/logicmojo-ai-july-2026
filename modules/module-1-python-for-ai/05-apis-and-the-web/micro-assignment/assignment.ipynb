## Problem 1: Read a JSON response

```python
# A response body, already parsed from JSON into a dict (what .json() gives you).
response = {"city": "Paris", "temp_c": 18, "conditions": "cloudy"}

# TODO: print "<city>: <temp_c>C, <conditions>".
# Then print the humidity if the key exists, else "n/a" (use .get with a default).

# your code here

print(f"{response.get('city')}: {response.get('temp_c')}C, {response.get('conditions')}")

humidity = response.get("humidity", "n/a")
print(f"humidity: {humidity}")
```

## Problem 2: Check the status code

```python
# Two responses, standing in for real requests results.
responses = [
    {"status_code": 200, "data": {"answer": 42}},
    {"status_code": 404, "data": None},
]

# TODO: for each response, if status_code is 200 print "ok: <data>",
# otherwise print "request failed with status <code>".

# your code here
for item in responses:
    if item.get("status_code") == 200:
        print(f"ok: {item.get('data')}")
    else:
        print(f"request failed with status {item.get('status_code')}")
```

## Problem 3: Call a local model with Ollama (run locally)

```python
import requests

# TODO: call your local Ollama model with a short prompt and print its reply.
# POST to http://localhost:11434/api/generate with
#   json={"model": "llama3.2", "prompt": "...", "stream": False}
# then read resp.json()["response"]. (Pull a small model first: ollama pull llama3.2)

# your code here
res = requests.post(
    "http://localhost:11434/api/generate",
    json={
        "model": "llama3.2",
        "prompt": "Say hello in one short sentence.",
        "stream": False,
    },
)

print(res.json()["response"])
```

## Problem 4: Call a hosted model with Groq (run locally)

```python
import os
import requests

# TODO: call Groq with the same prompt as Problem 3.
# Read the key from os.environ["GROQ_API_KEY"] (set it in your shell; never hardcode it).
# POST to https://api.groq.com/openai/v1/chat/completions with an Authorization header
# and a "messages" body, then read resp.json()["choices"][0]["message"]["content"].

# your code here
api_key = os.environ["GROQ_API_KEY"]

resp = requests.post(
    "https://api.groq.com/openai/v1/chat/completions",
    headers={
        "Authorization": f"Bearer {api_key}",
        "Content-Type": "application/json",
    },
    json={
        "model": "llama-3.1-8b-instant",
        "messages": [
            {"role": "user", "content": "Say hello in one short sentence."}
        ],
    },
)

print(resp.json()["choices"][0]["message"]["content"])
```
