## Problem 1: Read a JSON response

## Problem 2: Check the status code

```python
responses = [
    {"status_code": 200, "data": {"answer": 42}},
    {"status_code": 404, "data": None},
]

for r in responses:
    if r["status_code"] == 200:
        print(f"ok: {r['data']}")
    else:
        print(f"request failed with status {r['status_code']}")
```

## Problem 3: Call a local model with Ollama (run locally)

> Runs locally, so it is not executed in this reference notebook. Start Ollama and pull a small model first (`ollama pull llama3.2`). Your reply text will differ from the example in the README.

```python
import requests

# Needs Ollama running locally with a small model pulled: ollama pull llama3.2
resp = requests.post(
    "http://localhost:11434/api/generate",
    json={"model": "llama3.2", "prompt": "What's NATO? Give a brief explanation.", "stream": False},
)
reply = resp.json()["response"]
print(reply)
```

## Problem 4: Call a hosted model with Groq (run locally)

> Needs a free Groq API key in the `GROQ_API_KEY` environment variable and network access.

```python
import os
import requests
import getpass

# Set your key in the shell first, never in code:  export GROQ_API_KEY=your_key
# Get the API key from the user without echoing it to the console

if not os.environ.get('GROQ_API_KEY'):
    os.environ['GROQ_API_KEY'] = getpass.getpass("Enter your Groq API key: ")

resp = requests.post(
    "https://api.groq.com/openai/v1/chat/completions",
    headers={"Authorization": f"Bearer {os.environ['GROQ_API_KEY']}"},
    json={
        "model": "llama-3.1-8b-instant",   # check Groq's docs for a current model id
        "messages": [{"role": "user", "content": "What's BRICS? Give a brief explanation."}],
    },
)
reply = resp.json()["choices"][0]["message"]["content"]
print("groq:", reply)
```
