## Problem 1: An Account class

```
Account A balance: 150
Account B balance: 20
```

````python
# TODO: define a class Account with __init__(self, balance=0) storing self.balance
# and a deposit(self, amount) method. Create two accounts (start 100 and 0), deposit
# 50 and 20, and print both balances to show independent state.

# your code here
class Account:
    def __init__(self, balance=0):
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

a = Account(100)
a.deposit(50)
b = Account(0)
b.deposit(20)

print(f"Account A balance: {a.balance}")
print(f"Account B balance: {b.balance}")
````

## Problem 2: Sum the valid numbers

```
total: 20
skipped: 1
```

````python
raw = ["10", "7", "x", "3"]

# TODO: add up the items that are valid integers. Use try/except to catch the
# ValueError from int() on a bad item and skip it. Print the total and how many
# were skipped.

# your code here
total = 0
skipped = 0
for item in raw:
    try:
        num = int(item)
        total += num
    except ValueError:
        skipped += 1

print(f"total: {total}")
print(f"skipped: {skipped}")
````

## Problem 3: JSON is a dict in text form

```
as text: {"course": "GenAI", "classes": 45}
updated classes: 46
```

````python
import json

data = {"course": "GenAI", "classes": 45}

# TODO: turn the dict into JSON text with json.dumps, read it back with
# json.loads, add 1 to "classes", then print the JSON text and the new value.

# your code here
json_text = json.dumps(data)
reloaded = json.loads(json_text)
reloaded["classes"] += 1

print(f"as text: {json_text}")
print(f"updated classes: {reloaded['classes']}")
````

## Problem 4: A savings subclass

```
Account: 100
SavingsAccount: 200 at 0.05
```

````python
# TODO:
# 1. Define Account with __init__(self, balance) and summary() returning "Account: <balance>".
# 2. Define SavingsAccount(Account) that also takes a rate (call super().__init__ for the
#    balance) and overrides summary() to return "SavingsAccount: <balance> at <rate>".
# 3. Put one of each in a list and print each summary() (same call, different output).

# your code here
class Account:
    def __init__(self, balance):
        self.balance = balance

    def summary(self):
        return f"Account: {self.balance}"


class SavingsAccount(Account):
    def __init__(self, balance, rate):
        super().__init__(balance)
        self.rate = rate

    def summary(self):
        return f"SavingsAccount: {self.balance} at {self.rate}"


accounts = [Account(100), SavingsAccount(200, 0.05)]
for acc in accounts:
    print(acc.summary())
````

## Problem 5: Missing file, handled

```
could not find missing_data.json; starting empty
records: 0
```

````python
import json

filename = "missing_data.json"

# TODO: try to open and json.load this file. If it is missing, catch
# FileNotFoundError, print "could not find <filename>; starting empty", and use an
# empty list instead. Then print how many records you ended up with.

# your code here
try:
    with open(filename) as f:
        records = json.load(f)
except FileNotFoundError:
    print(f"could not find {filename}; starting empty")
    records = []

print(f"records: {len(records)}")
````
