## Problem 1: An Account class

```python
class Account:
    def __init__(self, balance=0):
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

a = Account(100)
a.deposit(50)
b = Account(0)
b.deposit(20)

print(f"Account A balance: {a.balance}")
print(f"Account B balance: {b.balance}")
```

## Problem 2: Sum the valid numbers

```python
raw = ["10", "7", "x", "3"]

total = 0
skipped = 0
for item in raw:
    try:
        total += int(item)
    except ValueError:
        skipped += 1

print(f"total: {total}")
print(f"skipped: {skipped}")
```

## Problem 3: JSON is a dict in text form

```python
import json

data = {"course": "GenAI", "classes": 45}
text = json.dumps(data)      # dict -> JSON text
back = json.loads(text)      # JSON text -> dict
back["classes"] = back["classes"] + 1

print(f"as text: {text}")
print(f"updated classes: {back['classes']}")
```

## Problem 4: A savings subclass

```python
class Account:
    def __init__(self, balance):
        self.balance = balance
    def summary(self):
        return f"Account: {self.balance}"

class SavingsAccount(Account):
    def __init__(self, balance, rate):
        super().__init__(balance)   # reuse the parent for balance
        self.rate = rate
    def summary(self):              # override
        return f"SavingsAccount: {self.balance} at {self.rate}"

accounts = [Account(100), SavingsAccount(200, 0.05)]
for acc in accounts:
    print(acc.summary())
```

## Problem 5: Missing file, handled

```python
import json

filename = "missing_data.json"
try:
    with open(filename) as f:
        data = json.load(f)
    print(f"loaded {len(data)} records")
except FileNotFoundError:
    print(f"could not find {filename}; starting empty")
    data = []

print(f"records: {len(data)}")
```
