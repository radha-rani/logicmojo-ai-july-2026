# Micro-assignment 1.4: Reference solution

One clear way to solve each problem, using only class 1.4 topics.

```python
import numpy as np
import pandas as pd
```

## Problem 1: Array basics

```python
arr = np.array([2, 4, 6, 8, 10])
print("plus 10:", arr + 10)

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
print("product:", a * b)

print("mean:", arr.mean())
```

## Problem 2: Above the mean

```python
data = np.array([3, 8, 1, 10, 6, 4])
m = data.mean()
above = data[data > m]
print(f"mean: {m:.2f}")
print("above mean:", above)
print("count:", above.size)
```

## Problem 3: Vectorized equals loop

```python
qty = np.array([2, 1, 3])
price = np.array([10, 25, 5])

total = np.sum(qty * price)          # vectorized: multiply then sum

loop_total = 0                       # the same thing with a loop
for i in range(len(qty)):
    loop_total += qty[i] * price[i]

print("vectorized:", total)
print("loop:", loop_total)
print("match:", total == loop_total)
```

## Problem 4: Normalize to 0..1

```python
scores = np.array([50, 20, 80, 40])
norm = (scores - scores.min()) / (scores.max() - scores.min())
print("normalized:", np.round(norm, 2))
```

## Problem 5: A DataFrame

```python
df = pd.DataFrame({
    "name": ["Ada", "Alan", "Grace", "Kay"],
    "score": [91, 72, 85, 60],
    "team": ["A", "B", "A", "B"],
})
print(df)
print()
print("scores:", list(df["score"]))
print()
print(df[df["score"] >= 85])
```

## Problem 6: Derived column and group

```python
df = pd.DataFrame({
    "name": ["Ada", "Alan", "Grace", "Kay"],
    "score": [91, 72, 85, 60],
    "team": ["A", "B", "A", "B"],
})
df["passed"] = df["score"] >= 70
print(df)
print()
print(df.groupby("team")["score"].mean())
```

## Problem 7: Clean a messy table

```python
raw = pd.DataFrame({
    "city": ["A", "B", "C", "D"],
    "sales": [100.0, np.nan, 250.0, 80.0],
    "visits": [10, 5, 25, 8],
})
clean = raw.dropna().copy()              # drop rows with a missing value
clean = clean[clean["sales"] > 90].copy()  # keep the bigger sales
clean["per_visit"] = clean["sales"] / clean["visits"]  # derive a column
print(clean)
```

## Problem 8: Top category

```python
sales = pd.DataFrame({
    "category": ["food", "toys", "food", "books", "toys"],
    "amount": [20, 15, 30, 10, 25],
})
totals = sales.groupby("category")["amount"].sum()
print(totals)
print()
print("top category:", totals.idxmax(), "with", totals.max())
```
