## Variables and types

A variable is a labeled box. A type is the kind of thing inside it.

````python
n = 7             # int
price = 4.5       # float
name = "Ada"      # str
is_ready = True   # bool

print(type(n), type(price), type(name), type(is_ready))
````

````python
# A variable can be reassigned, and its type follows the new value.
x = 10
print(x, type(x))

x = "now a string"
print(x, type(x))
````

## f-strings

Drop values straight into text. You will use this constantly.

````python
x = 54

print("The value of x is: ",x)
````

````python
count = 3
print(f"count is {count}")

name, score = "Ada", 91
print(f"{name} scored {score}")
````

````python
# You can format numbers inside the braces.
price = 4.565
print(f"total: ${price:.2f}")   # two decimal places
````

## Lists

A list is an ordered collection you can index, slice, and grow. Think of a numbered to-do list: items stay in order, you add to the bottom, and you read the nth line.

````python
names = ["Ada", "Alan", "Grace"]
print(names)
print(f"{len(names)} names")       # len() returns the number of items in a list
````

````python
# Index from 0; negative indexes count from the end.
print(names[0])    # first item
print(names[-1])   # last item
````

````python
# Grow the list
names.append("Kay")
print(names)
````

````python
# insert: add an item at a specific position in a list

l = ['a','b','c']

l.insert(2, 'x')

print(l)
````

````python
# Take slices of the list

print(names[:2])   # first two
print(names[1:3])  # a middle slice
````

````python
# Membership test: is a value in the list?
print("Ada" in names)
print("Zed" in names)
````

## Control flow: if, for, while

Decisions and repetition. The `for` loop below reuses the `names` list from above.

````python
# if: report the sign of a number
n = -4
if n > 0:
    print("positive")
elif n < 0:
    print("negative")
else:
    print("zero")
````

````python
n = 55

if n % 2 == 0:
    print("even", end=" ")
    if n % 3 == 0:
        print("and divisible by 3")
    else:
        print("and not divisible by 3")
else:
    print("odd")
````

````python
# for: do something once per item in the list

for person in names:   # names is an iterable. for loop works with any iterable, not just lists.
    print(person)
````

````python
# range() gives you a sequence of numbers
for i in range(1, 4):
    print(i)       # 1, 2, 3
````

````python
names
````

````python
# enumerate() adds a counter

for i, person in enumerate(names, start=1):
    print(f"{i}. {person}")
````

````python
# while: keep going until a condition stops being true
k = 3
while k > 0:
    print(k)
    k -= 1    # k = k-1
print("done")
````

## Reading a traceback

Errors are normal. Read the **last line first**: it names the error and what went wrong. Then look just above it for the line where it happened.

The next cell has a deliberate typo so you can practise. Run it, read the message, then run the fixed cell below. A missing colon or a wrong indent shows up as a `SyntaxError` or `IndentationError`, read the same way.

````python
# Deliberate error: 'namez' is a typo and was never defined.
for person in namez:
    print(person)
````

````python
# Fixed: use the name that actually exists.
for person in names:
    print(f"hello, {person}")
````
