## Problem 1: Names report

```
1. Ada
2. Alan
3. Grace (long)
4. Kay
5. Katherine (long)
6. Ed

2 of 6 names are longer than four letters.
```

````python
names = ["Ada", "Alan", "Grace", "Kay", "Katherine", "Ed"]

# TODO: number each name; add " (long)" if len(name) > 4;
# then print "<n> of <total> names are longer than four letters."

# your code here
count_long = 0
for i, name in enumerate(names,start=1):
    if len(name)>4:
        print(f"{i}. {name} (long)")
        count_long += 1
    else:
        print(f"{i}. {name}")

print(f"\n {count_long} of {len(names)} names are longer than four letters.")
````

## Problem 2: Reading signs

```
12: positive
-5: negative
0: zero
8: positive
-3: negative
7: positive

positives: 3
total: 19
```

````python
readings = [12, -5, 0, 8, -3, 7]

# TODO: print "value: sign" (positive/negative/zero) for each;
# then print "positives: <count>" and "total: <sum>".

# your code here
positives_count = 0
negatives_count = 0
total = 0
for num in readings:
    if num>0:
        print(f"{num}: positive")
        positives_count += 1
    elif num<0:
        print(f"{num}: negative")
        negatives_count += 1
    else:
        print(f"{num}: zero")

    total += num


print(f"\npositives: {positives_count}  \ntotal: {total}")
````

## Problem 3: Countdown list

```
[5, 4, 3, 2, 1]
length: 5
first: 5, last: 1
```

````python
n = 5

# TODO: use a while loop to build [5, 4, 3, 2, 1] by appending;
# then print the list, its length, and its first and last items.

# your code here

num_list = []
while n>0:
    num_list.append(n)
    n -= 1
first = num_list[0]
last = num_list[-1]
print(f"\n{num_list}\nlength:{len(num_list)}\nfirst: {first}, last: {last}")

````

## Problem 4: Highest reading, without max()

```
highest: 47
first seen at index: 1
appears 3 times
```

````python
scores = [12, 47, 47, 8, 5, 47, 19]

# TODO (no max()): find the highest value, the index where it first appears,
# and how many times it appears. Track the "highest so far" as you scan,
# then count appearances in a second pass.

# your code here
highest = scores[0]
first_seen = 0
for i, n in enumerate(scores):
    if n > highest:
        highest = n
        first_seen = i

count = 0
for n in scores:
    if n == highest:
        count += 1

print(f"highest: {highest}\nfirst seen at index: {first_seen}\nappears {count} times")

    


````

## Problem 5: Is it sorted?

```
sorted: False
first drop at index 3 (12 then 5)
```

````python
values = [3, 8, 8, 12, 5, 20]

# TODO: decide if the list is non-decreasing (each item <= the next).
# Compare values[i] with values[i + 1] over range(len(values) - 1).
# If it is not sorted, also report the index where the order first breaks.

# your code here
is_sorted = True
break_index = None

for i in range(len(values) - 1):
    if values[i] > values[i + 1]:
        is_sorted = False
        break_index = i
        break

print(f"sorted: {is_sorted}")
if not is_sorted:
    print(f"first drop at index {break_index} ({values[break_index]} then {values[break_index + 1]})")
````
