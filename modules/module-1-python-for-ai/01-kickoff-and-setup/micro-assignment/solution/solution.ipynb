# Micro-assignment 1.1: Reference solution

One clear way to solve each problem, using only class 1.1 topics (no functions, and no `break`/`and`/`not`).

## Problem 1: Names report

```python
names = ["Ada", "Alan", "Grace", "Kay", "Katherine", "Ed"]

long_count = 0
for i, name in enumerate(names, start=1):
    if len(name) > 4:
        long_count += 1
        print(f"{i}. {name} (long)")
    else:
        print(f"{i}. {name}")

print()
print(f"{long_count} of {len(names)} names are longer than four letters.")
```

## Problem 2: Reading signs

```python
readings = [12, -5, 0, 8, -3, 7]

positives = 0
total = 0
for value in readings:
    if value > 0:
        sign = "positive"
        positives += 1
    elif value < 0:
        sign = "negative"
    else:
        sign = "zero"
    total += value
    print(f"{value}: {sign}")

print()
print(f"positives: {positives}")
print(f"total: {total}")
```

## Problem 3: Countdown list

```python
n = 5

countdown = []
while n > 0:
    countdown.append(n)
    n -= 1

print(countdown)
print(f"length: {len(countdown)}")
print(f"first: {countdown[0]}, last: {countdown[-1]}")
```

## Problem 4: Highest reading, without max()

```python
scores = [12, 47, 47, 8, 5, 47, 19]

# Pass 1: track the highest value and where it first appears.
highest = scores[0]
first_i = 0
for i, value in enumerate(scores):
    if value > highest:
        highest = value
        first_i = i

# Pass 2: count how many times that highest value appears.
appearances = 0
for value in scores:
    if value == highest:
        appearances += 1

print(f"highest: {highest}")
print(f"first seen at index: {first_i}")
print(f"appears {appearances} times")
```

## Problem 5: Is it sorted?

```python
values = [3, 8, 8, 12, 5, 20]

is_sorted = True
first_drop = -1
for i in range(len(values) - 1):
    if values[i] > values[i + 1]:
        is_sorted = False
        if first_drop == -1:
            first_drop = i

print(f"sorted: {is_sorted}")
if first_drop != -1:
    print(f"first drop at index {first_drop} ({values[first_drop]} then {values[first_drop + 1]})")
```
