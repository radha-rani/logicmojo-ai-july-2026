### Problem 1: A similarity score (dot product)

```python
import numpy as np
q = np.array([2., 0., 1.]); k = np.array([1., 3., 1.])
# Print the dot-product score of q and k.
# your code here
score = np.dot(q, k)
print(f"score: {score}")
```

### Problem 2: Softmax over scores

```python
s = np.array([4., 1., 0.])
# Print the softmax weights (3 decimals) and their sum.
# your code here

exp_s = np.exp(s)
weights = exp_s / np.sum(exp_s)

print(f"weights: {np.round(weights, 3)}")
print(f"sum: {np.sum(weights)}")
```

### Problem 3: Scale by sqrt(d_k)

```python
scores = np.array([8., 4., 2.]); d_k = 4
# Print scores / sqrt(d_k).
# your code here

d_k = 4

scaled = scores / np.sqrt(d_k)
print(f"scaled: {scaled.tolist()}")
```

### Problem 4: The context vector (weighted blend)

```python
weights = np.array([0.9, 0.1]); values = np.array([[1., 0.], [0., 1.]])
# Print the weighted blend weights @ values.
# your code here

context = weights @ values
print(f"context: {context.tolist()}")
```

### Problem 5: Where does this token attend?

```python
row = np.array([0.1, 0.7, 0.2])
# Print the index of the most-attended token.
# your code here

most_attended = np.argmax(row)
print(f"most-attended index: {most_attended}")
```

### Problem 7: Causal mask (no peeking ahead)

A query at position 1 may attend to positions 0 and 1 only. Set positions 2 and 3 to `-inf`, softmax, and print the weights (three decimals).

```python
import numpy as np
scores = np.array([2.0, 1.0, 3.0, 0.5])
# set positions 2 and 3 to -inf, softmax, print np.round(weights, 3)
# your code here

scores = np.array([2.0, 1.0, 3.0, 0.5])

masked_scores = scores.copy()
masked_scores[2:] = -np.inf

exp_scores = np.exp(masked_scores)
weights = exp_scores / np.sum(exp_scores)

print(f"weights: {np.round(weights, 3)}")
```
