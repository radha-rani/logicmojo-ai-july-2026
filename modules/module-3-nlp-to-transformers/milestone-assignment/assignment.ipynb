### Part 1. Tokenization audit (class 3.1, Text as data)

Load the `bert-base-uncased` tokenizer. For each message print the word count, the token count, and the token pieces. Then print the pieces for a few domain words and note which ones split.

````python
from transformers import AutoTokenizer

# 1. Load the "bert-base-uncased" tokenizer.
# 2. For each message, print its word count, token count, and tok.tokenize(message).
# 3. Print tok.tokenize(w) for a few domain words (e.g. reconciliation, invoices, csv).
# 4. In a comment, say why token count matters more than word count.
# your code here


tok = AutoTokenizer.from_pretrained("bert-base-uncased")

messages = [
    "I was charged twice for my subscription this month and I want a refund.",
    "My card was declined at checkout but I was still billed for the upgrade.",
    "The app crashes every time I open the bank reconciliation screen.",
    "After the latest update my reports show the wrong totals, this is broken.",
    "How do I export my invoices to a CSV file?",
    "Where is the setting to add a second user to my account?",
    "Please cancel my subscription, I do not need the service any more.",
    "I want to close my account and stop all future payments.",
]

for msg in messages:
    word_count = len(msg.split())
    tokens = tok.tokenize(msg)
    token_count = len(tokens)
    print(f"words: {word_count} | tokens: {token_count} | {tokens}")

print()
domain_words = ["reconciliation", "invoices", "csv"]
for w in domain_words:
    print(f"{w}: {tok.tokenize(w)}")

# Token count matters more than word count because models are billed and
# rate-limited by tokens, not words. A message with domain-specific or unusual
# words can silently use up more of the token budget than its word count suggests.
````

### Part 2. Sentiment triage (class 3.4, Using pretrained transformers)

Score every message with the sentiment pipeline, flag confident negatives, and sort so the angriest come first.

````python
from transformers import pipeline

# 1. Build a "sentiment-analysis" pipeline and score all messages.
# 2. Flag a message HIGH priority when it is NEGATIVE with score > 0.9.
# 3. Sort so the most negative print first.
# 4. In a comment, name the limitation this model shows on the how-to questions.
# your code here

sentiment_pipe = pipeline("sentiment-analysis")

results = []
for msg in messages:
    result = sentiment_pipe(msg)[0]
    results.append({
        "message": msg,
        "label": result["label"],
        "score": result["score"],
        "priority": "HIGH" if result["label"] == "NEGATIVE" and result["score"] > 0.9 else "normal"
    })

# sort so the most negative (lowest confidence-adjusted, i.e. most confidently NEGATIVE) print first
results_sorted = sorted(results, key=lambda r: r["score"] if r["label"] == "NEGATIVE" else -r["score"], reverse=True)

for r in results_sorted:
    print(f"[{r['priority']}] {r['label']} ({r['score']:.3f}): {r['message']}")

# Limitation: this SST-2 model only outputs POSITIVE or NEGATIVE, no NEUTRAL class.
# In this run, it didn't just mislabel the neutral how-to questions -- it labeled
# them NEGATIVE with near-maximum confidence (0.999-1.000), just as confidently as
# genuine complaints. This shows the risk isn't only misclassification; a binary
# sentiment model can be fully confident while being wrong about whether a message
# is emotionally negative at all. Confidence score here reflects certainty between
# the two available labels, not certainty that either label actually applies.
````

### Part 3. Routing and FAQ retrieval with embeddings (classes 3.2, 3.3, 3.4)

First mean-pool a DistilBERT `last_hidden_state` to compare a paraphrase pair against an unrelated sentence. Then use a sentence-embedding model to route every message to the nearest team, and retrieve the best FAQ for the two how-to questions.

````python
from transformers import AutoTokenizer, AutoModel
import torch
import torch.nn.functional as F

# 1. Write embed_meanpool(text): run distilbert-base-uncased, mean-pool the
#    last_hidden_state using the attention mask, return one vector.
# 2. Show cosine(paraphrase pair) is higher than cosine(unrelated pair).
# your code here


tok2 = AutoTokenizer.from_pretrained("distilbert-base-uncased")
model2 = AutoModel.from_pretrained("distilbert-base-uncased")

def embed_meanpool(text):
    inputs = tok2(text, return_tensors="pt")
    with torch.no_grad():
        outputs = model2(**inputs)
    hidden = outputs.last_hidden_state  # shape: (1, seq_len, hidden_dim)
    mask = inputs["attention_mask"].unsqueeze(-1).float()  # shape: (1, seq_len, 1)
    summed = (hidden * mask).sum(dim=1)
    counts = mask.sum(dim=1)
    mean_pooled = summed / counts
    return mean_pooled.squeeze(0)

# a paraphrase pair (same meaning, different wording)
sent_a = "Please cancel my subscription, I do not need the service any more."
sent_b = "I want to close my account and stop all future payments."

# an unrelated pair
sent_c = "The app crashes every time I open the bank reconciliation screen."

vec_a = embed_meanpool(sent_a)
vec_b = embed_meanpool(sent_b)
vec_c = embed_meanpool(sent_c)

sim_paraphrase = F.cosine_similarity(vec_a.unsqueeze(0), vec_b.unsqueeze(0)).item()
sim_unrelated = F.cosine_similarity(vec_a.unsqueeze(0), vec_c.unsqueeze(0)).item()

print(f"paraphrase pair similarity: {sim_paraphrase:.3f}")
print(f"unrelated pair similarity: {sim_unrelated:.3f}")
````

````python
from sentence_transformers import SentenceTransformer, util

# 3. Load "all-MiniLM-L6-v2". Encode the messages and the four team descriptions
#    (list(teams.values())) with normalize_embeddings=True.
# 4. Route each message to the team with the highest cosine similarity and print it.
# your code here
from sentence_transformers import SentenceTransformer, util

embedder = SentenceTransformer("all-MiniLM-L6-v2")

team_names = list(teams.keys())
team_descriptions = list(teams.values())

message_embeddings = embedder.encode(messages, normalize_embeddings=True)
team_embeddings = embedder.encode(team_descriptions, normalize_embeddings=True)

similarities = util.cos_sim(message_embeddings, team_embeddings)

for i, msg in enumerate(messages):
    best_team_idx = similarities[i].argmax().item()
    best_score = similarities[i][best_team_idx].item()
    print(f"{msg}\n  -> routed to: {team_names[best_team_idx]} (similarity: {best_score:.3f})\n")
````

````python
# 5. Encode the faqs. For messages 5 and 6 (the how-to questions), print the
#    single best-matching FAQ entry.
# your code here

faq_embeddings = embedder.encode(faqs, normalize_embeddings=True)

howto_messages = [messages[4], messages[5]]  # message 5 = index 4, message 6 = index 5

howto_embeddings = embedder.encode(howto_messages, normalize_embeddings=True)
faq_similarities = util.cos_sim(howto_embeddings, faq_embeddings)

for i, msg in enumerate(howto_messages):
    best_faq_idx = faq_similarities[i].argmax().item()
    best_score = faq_similarities[i][best_faq_idx].item()
    print(f"{msg}\n  -> best FAQ: {faqs[best_faq_idx]} (similarity: {best_score:.3f})\n")
````

### Part 4. Draft a reply and control the decoding (class 3.5, From models to LLMs)

Use `distilgpt2` to draft a reply to message 5, three ways: greedy, high temperature, and top-p. Print the prompt token count against the context window.

````python
from transformers import AutoTokenizer, AutoModelForCausalLM, set_seed
import torch

# 1. Load distilgpt2 (tokenizer + AutoModelForCausalLM).
# 2. Build a prompt from message 5 and print prompt tokens vs config.n_positions.
# 3. Generate 40 new tokens three times from the same prompt, calling set_seed(0)
#    before each: greedy (do_sample=False); temperature=1.3, top_k=0; top_p=0.9, top_k=0.
# 4. In a comment, say how the three drafts differ.
# your code here



gen_tok = AutoTokenizer.from_pretrained("distilgpt2")
gen_model = AutoModelForCausalLM.from_pretrained("distilgpt2")

prompt = f"Customer question: {messages[4]}\nSupport reply:"

prompt_ids = gen_tok(prompt, return_tensors="pt")
prompt_token_count = prompt_ids["input_ids"].shape[1]
context_window = gen_model.config.n_positions

print(f"prompt tokens: {prompt_token_count}")
print(f"context window: {context_window}")
print(f"fraction used: {prompt_token_count / context_window:.4f}")
print()

# Greedy decoding
set_seed(0)
greedy_out = gen_model.generate(
    **prompt_ids,
    max_new_tokens=40,
    do_sample=False,
    pad_token_id=gen_tok.eos_token_id
)
print("GREEDY:")
print(gen_tok.decode(greedy_out[0], skip_special_tokens=True))
print()

# High temperature sampling
set_seed(0)
hightemp_out = gen_model.generate(
    **prompt_ids,
    max_new_tokens=40,
    do_sample=True,
    temperature=1.3,
    top_k=0,
    pad_token_id=gen_tok.eos_token_id
)
print("HIGH TEMPERATURE (1.3):")
print(gen_tok.decode(hightemp_out[0], skip_special_tokens=True))
print()

# Top-p (nucleus) sampling
set_seed(0)
topp_out = gen_model.generate(
    **prompt_ids,
    max_new_tokens=40,
    do_sample=True,
    top_p=0.9,
    top_k=0,
    pad_token_id=gen_tok.eos_token_id
)
print("TOP-P (0.9):")
print(gen_tok.decode(topp_out[0], skip_special_tokens=True))

# Comparing the three: greedy is deterministic and tends to repeat the same
# phrase or loop, since it always picks the single highest-probability token
# (Micro-assignment 3.5, Problem 4 -- this is exactly argmax on the logits).
# The high-temperature sample is noticeably more varied and can wander off-topic,
# since flattening the distribution (Problem 2) gives real probability to much
# less likely words. Top-p sits in between: fluent and on-topic, because it only
# samples from the small nucleus of genuinely plausible next tokens (Problem 3),
# avoiding both greedy's repetition and high-temperature's incoherence.
````
