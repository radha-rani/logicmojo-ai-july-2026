### Problem 1: Whitespace tokenize and count

```python
s = "Attention is all you need"
# Lowercase, split on spaces; print the tokens and the count.
# your code here

tokens = s.lower().split()

print(f"tokens: {tokens}")
print(f"count: {len(tokens)}")
```

### Problem 3: Encode to ids and decode back

```python
vocab = {"i": 0, "love": 1, "nlp": 2}
tokens = ["i", "love", "nlp", "love"]
# Encode tokens to ids, then decode the ids back to tokens; print both.
# your code here



ids = [vocab[token] for token in tokens]

id_to_token = {v: k for k, v in vocab.items()}
decoded = [id_to_token[i] for i in ids]

print(f"ids: {ids}")
print(f"decoded: {decoded}")
```

### Problem 4: Token count is not word count

```python
text = "cats running"
tokens = ["cat", "##s", "runn", "##ing"]
# Print the word count and the token count.
# your code here

word_count = len(text.split())
token_count = len(tokens)

print(f"words: {word_count}")
print(f"tokens: {token_count}")
```

### Problem 5: A bag of tokens loses order

```python
a = "dog bites man"
b = "man bites dog"
# Print whether the two have the same bag of tokens, and whether they are the same sentence.
# your code here

tokens_a = a.split()
tokens_b = b.split()

same_bag = sorted(tokens_a) == sorted(tokens_b)
same_sentence = a == b

print(f"same bag: {same_bag}")
print(f"same sentence: {same_sentence}")
```
