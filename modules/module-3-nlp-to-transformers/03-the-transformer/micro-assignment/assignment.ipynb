### Problem 1: Feed-forward parameter count

```python
d = 8
# A feed-forward layer is Linear(d, 4d) then Linear(4d, d), each with weights and biases.
# Print the total parameter count.
# your code here

layer1_params = (d * 4*d) + 4*d  # weights + bias
layer2_params = (4*d * d) + d    # weights + bias

total_params = layer1_params + layer2_params
print(f"feed-forward params: {total_params}")
```
