### Problem 1: Softmax with temperature

```python
import numpy as np
# Write softmax_T(logits, T) = softmax(logits / T). Print the probs for [2,1,0] at T=1 (3 decimals).
# your code here
def softmax_T(logits, T):
    scaled = np.array(logits) / T
    exp_scaled = np.exp(scaled)
    return exp_scaled / np.sum(exp_scaled)

probs = softmax_T([2, 1, 0], 1)
print(f"probs: {np.round(probs, 3)}")
```

### Problem 2: Lower temperature is sharper

```python
# Print the max probability of [2,1,0] at T=0.5 and at T=2 (2 decimals).
# your code here
def softmax_T(logits, T):
    scaled = np.array(logits) / T
    exp_scaled = np.exp(scaled)
    return exp_scaled / np.sum(exp_scaled)

probs_low = softmax_T([2, 1, 0], 0.5)
probs_high = softmax_T([2, 1, 0], 2)

print(f"T=0.5 max: {round(max(probs_low), 2)}")
print(f"T=2   max: {round(max(probs_high), 2)}")
```

### Problem 3: Nucleus size for top-p

```python
probs = [0.5, 0.25, 0.15, 0.1]
# Print how many tokens are in the nucleus for p=0.9.
# your code here

p = 0.9

running_total = 0
nucleus_size = 0

for prob in probs:
    running_total += prob
    nucleus_size += 1
    if running_total >= p:
        break

print(f"nucleus size (p=0.9): {nucleus_size}")
```
