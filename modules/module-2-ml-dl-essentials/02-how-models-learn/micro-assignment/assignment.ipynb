### Problem 1: Mean squared error by hand

```python
y_hat = [4.0, 1.0, 3.0, 6.0]
y     = [3.5, 2.0, 3.0, 5.0]
# Compute the MSE and print it to two decimals.
# your code here
squared_errors = []
for pred, actual in zip(y_hat, y):
    error = pred - actual
    squared_errors.append(error ** 2)
mse = sum(squared_errors)/len(squared_errors)
print(f"MSE: {round(mse, 2)}")
```

### Problem 3: A five-step descent

```python
# Minimize f(x) = (x-3)**2 (slope 2*(x-3)) from x=0.0, lr=0.3.
# Take five steps; print the list of x values (start first), each rounded to 3 decimals.
# your code here
# x each step: [0.0, 1.8, 2.52, 2.808, 2.923, 2.969]
x = 0.0
lr = 0.3
x_values = [round(x, 3)]

for step in range(5):
    slope = 2*(x-3)
    x = x - lr * slope
    x_values.append(round(x, 3))

print(f"x each step: {x_values}")
```

### Problem 4: Diagnose three learning rates

```python
# Minimize f(x) = x**2 from x=1.0 for 15 steps at lr 0.05, 0.4, 1.1.
# Label each "diverges" (final loss higher than start), "converges" (final < 0.01), else "crawls".
# Print the list of three labels.
# your code here
x = 1.0
lrs = [0.05, 0.4, 1.1]
labels = []

for lr in lrs:
    x_current = x
    start_loss = x_current ** 2
    for step in range(15):
        slope = 2 * x_current
        x_current = x_current - lr * slope
    final_loss = x_current ** 2

    if final_loss > start_loss:
        labels.append("diverges")
    elif final_loss < 0.01:
        labels.append("converges")
    else:
        labels.append("crawls")

print(labels)
```

### Problem 6: Cross-entropy on a confident-wrong guess

```python
# True label is 1, so cross-entropy is -log(p). Compute it for p=0.9 and p=0.1 (two decimals),
# then print one line on what the gap shows.
# your code here
import math

p_right = 0.9
p_wrong = 0.1

ce_right = -math.log(p_right)
ce_wrong = -math.log(p_wrong)

print(f"cross-entropy at p=0.9: {ce_right:.2f}")
print(f"cross-entropy at p=0.1: {ce_wrong:.2f}")
print("note: the confident-wrong guess costs far more, which is why classification uses cross-entropy")
```
