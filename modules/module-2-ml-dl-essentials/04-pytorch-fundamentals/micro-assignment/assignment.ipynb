### Problem 4: The five-line loop lowers the loss

```python
x = torch.tensor([1., 2., 3., 4.])
y = torch.tensor([2., 4., 6., 8.])
w = torch.tensor([0.0], requires_grad=True)
# Fit w so w*x matches y: MSE loss, SGD(lr=0.02), 80 steps (the five-line loop).
# Print the first-step loss, the final loss, and whether it decreased.
# your code here


loss_fn = nn.MSELoss()
opt = torch.optim.SGD([w], lr=0.02)

losses = []
for step in range(80):
    pred = w * x
    loss = loss_fn(pred, y)
    opt.zero_grad()
    loss.backward()
    opt.step()
    losses.append(loss.item())

first_loss = losses[0]
last_loss = losses[-1]

print(f"first loss: {round(first_loss, 2)}")
print(f"last loss: {round(last_loss, 2)}")
print(f"decreased: {last_loss < first_loss}")
```

### Problem 5: The zero_grad trap

```python
# Start w = torch.tensor([1.0], requires_grad=True). Call (w**2).sum().backward() three
# times without resetting and record w.grad each time; then repeat resetting with w.grad.zero_().
# Print both lists.
# your code here
# Part 1: without resetting
w = torch.tensor([1.0], requires_grad=True)
grads_without_reset = []

for i in range(3):
    (w ** 2).sum().backward()
    grads_without_reset.append(w.grad.item())

# Part 2: with resetting
w = torch.tensor([1.0], requires_grad=True)
grads_with_reset = []

for i in range(3):
    if w.grad is not None:
        w.grad.zero_()
    (w ** 2).sum().backward()
    grads_with_reset.append(w.grad.item())

print(f"without zero_grad: {grads_without_reset}")
print(f"with zero_grad: {grads_with_reset}")
```
