### Problem 1: A neuron forward pass

```python
import numpy as np
w = np.array([0.5, -0.2, 0.1]); x = np.array([2, 1, 4]); b = 0.3
# Compute z, then ReLU(z), print both to two decimals.
# your code here

z = np.dot(w, x) + b
relu_z = max(0, z)

print(f"z: {round(z, 2)}")
print(f"ReLU(z): {round(relu_z, 2)}")
```

### Problem 2: Apply ReLU, sigmoid, and tanh

```python
v = np.array([-2.0, 0.0, 3.0])
# Apply ReLU, sigmoid, and tanh, print each rounded to two decimals.
# your code 

relu_v = np.maximum(0, v)
sigmoid_v = 1 / (1 + np.exp(-v))
tanh_v = np.tanh(v)

print(f"ReLU: {np.round(relu_v, 2)}")
print(f"sigmoid: {np.round(sigmoid_v, 2)}")
print(f"tanh: {np.round(tanh_v, 2)}")
```

### Problem 3: A layer of two neurons

```python
W = np.array([[0.5, -1.0], [1.0, 0.5]])
b = np.array([0.0, 1.0])
x = np.array([2.0, 3.0])
# Compute W @ x + b, apply ReLU, print the two outputs.
# your code here

z = W @ x + b
outputs = np.maximum(0, z)

print(f"outputs: {outputs}")
```

### Problem 4: Two linear layers collapse to one

```python
W1 = np.array([[1.0, 2.0], [0.0, 1.0]]); b1 = np.array([0.0, 0.0])
W2 = np.array([[2.0, -1.0]]);            b2 = np.array([0.0])
x = np.array([1.0, 1.0])
# Compute the output through both layers, then through W2 @ W1, and print both plus whether they match.
# your code here
layer1_out = W1 @ x + b1
layer2_out = W2 @ layer1_out + b2

combined_out = (W2 @ W1) @ x + b2

print(f"two layers: {layer2_out}")
print(f"one layer: {combined_out}")
print(f"identical: {np.allclose(layer2_out, combined_out)}")
```

### Problem 5: Softmax

```python
logits = np.array([1.0, 2.0, 3.0])
# Print the softmax probabilities (3 decimals) and their sum.
# your code here
exp_logits = np.exp(logits)
probs = exp_logits / np.sum(exp_logits)

print(f"probs: {np.round(probs, 3)}")
print(f"sum: {np.sum(probs)}")
```

### Problem 6: Cross-entropy against a one-hot target

```python
y = np.array([0, 0, 1])            # one-hot target, true class is index 2
probs = np.array([0.1, 0.2, 0.7])
# Compute -sum(y * log(p)) to two decimals, confirm it equals -log(p_true), print both,
# then one line on why confident-wrong costs most.
# your code here
cross_entropy = -np.sum(y * np.log(probs))
p_true = probs[2]
neg_log_p_true = -np.log(p_true)

print(f"cross-entropy: {round(cross_entropy, 2)}")
print(f"same as -log(p_true): {round(neg_log_p_true, 2)}")
print("note: cross-entropy grows toward infinity as the probability on the true class nears 0, so a confident wrong answer is punished hardest")
```
