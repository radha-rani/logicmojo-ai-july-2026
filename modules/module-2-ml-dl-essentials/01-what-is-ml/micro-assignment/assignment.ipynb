### Problem 2: Three-way split and model selection

```python
# Load load_wine. Split off 20% test (random_state=0, stratify=y), then split the rest
# into train/validation (test_size=0.25, random_state=0, stratify). Scale on train only.
# Train LogisticRegression(max_iter=5000) and DecisionTreeClassifier(max_depth=2, random_state=0),
# pick the higher validation accuracy, and print sizes, both validation accuracies, chosen, test accuracy.
# your code here
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier

X,y = load_wine(return_X_y=True)

# Step 1- split off test (20%)

X_temp, X_test, y_temp, y_test = train_test_split( X, y, test_size=0.20, random_state=0, stratify=y)

# Step 2 - split the remaining 80% into train/validation

X_train, X_val, y_train, y_val = train_test_split(X_temp, y_temp, test_size=0.25, random_state=0, stratify=y_temp)

# Step 3: scale using train only
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_val_scaled = scaler.transform(X_val)
X_test_scaled = scaler.transform(X_test)

# Step 4: LogisticRegression
logreg = LogisticRegression(max_iter=5000)
logreg.fit(X_train_scaled, y_train)
val_acc_logreg = logreg.score(X_val_scaled, y_val)

# Step 4: DecisionTreeClassifier
dt_classifier = DecisionTreeClassifier(max_depth=2, random_state=0)
dt_classifier.fit(X_train_scaled, y_train)
val_acc_dt_classifier = dt_classifier.score(X_val_scaled, y_val)

# Step 5: pick the better one, check it on test

if val_acc_logreg >= val_acc_dt_classifier:
    chosen_model = logreg
else:
    chosen_model = dt_classifier

test_accuracy = chosen_model.score(X_test_scaled, y_test)

print(f"sizes train/val/test: {len(X_train)} {len(X_val)} {len(X_test)}")
print(f"validation logreg: {round(val_acc_logreg, 2)}")
print(f"validation tree: {round(val_acc_dt_classifier, 2)}")
print(f"test accuracy: {round(test_accuracy, 2)}")
```

### Problem 6: Precision and recall on imbalanced labels

```python
labels = [0] * 18 + [1, 1]
preds  = [0] * 20
# Treat 1 as positive. Compute accuracy, precision, recall by hand (precision 0.0 if no positive preds).
# Print the three numbers and one line on why accuracy flatters this model.
# your code here


tp = fp = fn = tn = 0
for actual, predicted in zip(labels, preds):
    if predicted == 1 and actual == 1:
        tp += 1
    elif predicted == 1 and actual == 0:
        fp += 1
    elif predicted == 0 and actual == 1:
        fn += 1
    else:
        tn += 1

accuracy = (tp + tn) / len(labels)
precision = tp / (tp + fp) if (tp + fp) > 0 else 0.0
recall = tp / (tp + fn) if (tp + fn) > 0 else 0.0

print(f"accuracy: {round(accuracy, 2)}")
print(f"precision: {round(precision, 2)}")
print(f"recall: {round(recall, 2)}")
print("note: accuracy is high, but recall 0.0 means it caught none of the rare positives")
```
